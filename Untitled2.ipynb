```python
data_str = input("Enter numbers separated by spaces: ")
numbers = []
for s in data_str.split():
    try:
        numbers.append(float(s))
    except ValueError:
        print(f"Skipping invalid input: {s}")

# Ensure we have some numbers to work with
if not numbers:
    print("No valid numbers entered. Please run again and enter numbers.")
else:
    print(f"You entered: {numbers}")
```

```python
if numbers:
    total_sum = 0
    count = 0
    for num in numbers:
        total_sum += num
        count += 1

    mean = total_sum / count
    print(f"Mean: {mean}")
```

```python
if numbers:
    # Sort the list (creating a copy to not alter original order if needed later)
    sorted_numbers = []
    # Manual bubble sort for demonstration, to avoid `list.sort()` or `sorted()`
    temp_numbers = list(numbers) # Create a mutable copy
    n = len(temp_numbers)
    for i in range(n):
        for j in range(0, n-i-1):
            if temp_numbers[j] > temp_numbers[j+1]:
                temp_numbers[j], temp_numbers[j+1] = temp_numbers[j+1], temp_numbers[j]
    sorted_numbers = temp_numbers

    n = len(sorted_numbers)
    if n % 2 == 1: # Odd number of elements
        median = sorted_numbers[n // 2]
    else: # Even number of elements
        mid1 = sorted_numbers[n // 2 - 1]
        mid2 = sorted_numbers[n // 2]
        median = (mid1 + mid2) / 2
    print(f"Median: {median}")
```

```python
if numbers:
    # We need the mean from the previous step
    # If the previous cell was not run, 'mean' might not be defined.
    # For robustness, we re-calculate mean here if it's not present
    try:
        mean
    except NameError:
        total_sum = 0
        count = 0
        for num in numbers:
            total_sum += num
            count += 1
        mean = total_sum / count

    squared_differences_sum = 0
    for num in numbers:
        diff = num - mean
        squared_differences_sum += diff * diff

    # Population variance (divide by n)
    variance = squared_differences_sum / len(numbers)
    print(f"Variance: {variance}")
```

```python
if numbers:
    # We need variance from the previous step
    # For robustness, re-calculate variance if not present
    try:
        variance
    except NameError:
        # Re-calculate mean first
        try:
            mean
        except NameError:
            total_sum = 0
            count = 0
            for num in numbers:
                total_sum += num
                count += 1
            mean = total_sum / count

        squared_differences_sum = 0
        for num in numbers:
            diff = num - mean
            squared_differences_sum += diff * diff
        variance = squared_differences_sum / len(numbers)

    # Manual square root (Newton's method or just using **0.5)
    # Using **0.5 as it's an operator, not a 'function' like math.sqrt()
    standard_deviation = variance ** 0.5
    print(f"Standard Deviation: {standard_deviation}")
```
