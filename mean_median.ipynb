```python


# Input dataset
data = [10, 20, 30, 40, 50]

print("Dataset:", data)

# Number of elements
n = len(data)

# Mean
sum_data = sum(data)
mean = sum_data / n
print("Mean =", mean)

# Median
data.sort()
if n % 2 == 0:
    median = (data[n//2 - 1] + data[n//2]) / 2
else:
    median = data[n//2]
print("Median =", median)

# Variance
variance_sum = 0
for i in data:
    variance_sum = variance_sum + (i - mean) ** 2

variance = variance_sum / n
print("Variance =", variance)

# Standard Deviation
std_dev = variance ** 0.5
print("Standard Deviation =", std_dev)
```
