```python
import numpy as np

# Define the coefficient matrix A
A = np.array([[2, 3],
              [4, 5]])

# Define the constant vector B
B = np.array([10, 20])

# Solve the system of equations
x, y = np.linalg.solve(A, B)

print(f"The solution is: x = {x:.2f}, y = {y:.2f}")
```
