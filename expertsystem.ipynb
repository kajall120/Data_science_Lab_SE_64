```python
print("Welcome to the Simple Car Diagnostic Expert System!")
print("Please answer the following questions with 'yes' or 'no'.")

problem = input("Is the car not starting? (yes/no): ").lower()

if problem == 'yes':
    lights_on = input("Are the lights coming on? (yes/no): ").lower()

    if lights_on == 'no':
        print("Diagnosis: The battery might be dead or there's a loose battery connection. Check the battery.")
    else:
        cranking = input("Is the engine cranking but not starting? (yes/no): ").lower()

        if cranking == 'yes':
            print("Diagnosis: Could be a fuel delivery issue (empty tank, fuel pump) or a spark plug problem.")
        else:
            print("Diagnosis: It might be a starter motor issue or an electrical problem.")
else:
    print("Diagnosis: If the car is starting, what is the specific problem you are experiencing?")
    print("This system is designed for 'car not starting' issues. Please provide more details for other problems.")
```
