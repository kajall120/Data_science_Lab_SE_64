```python
# Equations
print("Given equations:")
print("2x + 3y = 10")
print("4x + 5y = 20")

# Coefficients
a1, b1, c1 = 2, 3, 10
a2, b2, c2 = 4, 5, 20

# Determinant
D = a1*b2 - a2*b1

# Solve
x = (c1*b2 - c2*b1) / D
y = (a1*c2 - a2*c1) / D

print("\nSolution:")
print("x =", x)
print("y =", y)
```
