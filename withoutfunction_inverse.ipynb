```python

A = [
    [1, 2, 0],
    [0, 1, 0],
    [1, 0, 1]
]

# Function to find determinant of 3x3 matrix
def determinant(m):
    return (
        m[0][0]*(m[1][1]*m[2][2] - m[1][2]*m[2][1])
        - m[0][1]*(m[1][0]*m[2][2] - m[1][2]*m[2][0])
        + m[0][2]*(m[1][0]*m[2][1] - m[1][1]*m[2][0])
    )

# Function to find cofactor matrix
def cofactor(m):
    cof = [[0]*3 for _ in range(3)]

    cof[0][0] =  (m[1][1]*m[2][2] - m[1][2]*m[2][1])
    cof[0][1] = -(m[1][0]*m[2][2] - m[1][2]*m[2][0])
    cof[0][2] =  (m[1][0]*m[2][1] - m[1][1]*m[2][0])

    cof[1][0] = -(m[0][1]*m[2][2] - m[0][2]*m[2][1])
    cof[1][1] =  (m[0][0]*m[2][2] - m[0][2]*m[2][0])
    cof[1][2] = -(m[0][0]*m[2][1] - m[0][1]*m[2][0])

    cof[2][0] =  (m[0][1]*m[1][2] - m[0][2]*m[1][1])
    cof[2][1] = -(m[0][0]*m[1][2] - m[0][2]*m[1][0])
    cof[2][2] =  (m[0][0]*m[1][1] - m[0][1]*m[1][0])

    return cof


def transpose(m):
    return [[m[j][i] for j in range(3)] for i in range(3)]


det = determinant(A)

if det == 0:
    print("Matrix has no inverse")
else:
    cof = cofactor(A)
    adj = transpose(cof)

    # Inverse = adjoint / determinant
    inverse = [[adj[i][j]/det for j in range(3)] for i in range(3)]

    print("Inverse Matrix:")
    for row in inverse:
        print(row)
```
