```python
class Node:
    def __init__(self, name, value=None, children=None):
        self.name = name
        self.value = value  # Only for leaf nodes
        self.children = children if children is not None else []

    def is_leaf(self):
        return not self.children

    def __repr__(self):
        return f"Node('{self.name}', value={self.value})"

def minimax(node, maximizing_player):
    if node.is_leaf():
        return node.value

    if maximizing_player:
        max_eval = -float('inf')
        for child in node.children:
            eval = minimax(child, False)
            max_eval = max(max_eval, eval)
        return max_eval
    else:
        min_eval = float('inf')
        for child in node.children:
            eval = minimax(child, True)
            min_eval = min(min_eval, eval)
        return min_eval

# Define the graph structure with placeholder leaf values
# Leaves
D = Node('D', value=4)
E = Node('E', value=7)
F = Node('F', value=2)
G = Node('G', value=9)
H = Node('H', value=1)
I = Node('I', value=5)

# Intermediate nodes
B = Node('B', children=[D, E, F])
C = Node('C', children=[G, H, I])

# Root node
A = Node('A', children=[B, C])

print("Graph Structure (from root A):")
print(f"  Children of A: {A.children}")
print(f"    Children of B: {B.children}")
print(f"    Children of C: {C.children}")

# Run Minimax algorithm, assuming the root player (A) is the maximizing player
optimal_value = minimax(A, True)
print(f"\nThe optimal value for the root node A (maximizing player) is: {optimal_value}")
```

### Explanation of the Minimax Algorithm Implementation:

1.  **`Node` Class**: This class represents each node in the game tree. Each node has:
    *   `name`: A string identifier (e.g., 'A', 'B', 'D').
    *   `value`: An integer representing the utility of the node. This is only relevant for *leaf* nodes (terminal states of the game).
    *   `children`: A list of `Node` objects that are direct descendants of this node.
    *   `is_leaf()`: A helper method to check if a node has no children.

2.  **`minimax(node, maximizing_player)` Function**: This is the core recursive function that implements the Minimax algorithm.
    *   **`node`**: The current node being evaluated.
    *   **`maximizing_player`**: A boolean indicating whose turn it is. `True` if it's the maximizing player (trying to get the highest score), `False` if it's the minimizing player (trying to get the lowest score).

    *   **Base Case**: If `node.is_leaf()` is true, it means we've reached a terminal state. The function simply returns the `node.value`.

    *   **Recursive Step (Maximizing Player)**:
        *   If `maximizing_player` is `True`, the current player wants to maximize their score. It initializes `max_eval` to negative infinity.
        *   It then recursively calls `minimax` for each `child` node, but this time with `maximizing_player` set to `False` (because it will be the minimizing player's turn next).
        *   It updates `max_eval` to the maximum of the current `max_eval` and the `eval` returned from the child's call.

    *   **Recursive Step (Minimizing Player)**:
        *   If `maximizing_player` is `False`, the current player wants to minimize the opponent's score (which means minimizing their own score). It initializes `min_eval` to positive infinity.
        *   It recursively calls `minimax` for each `child` node, with `maximizing_player` set to `True` (it will be the maximizing player's turn next).
        *   It updates `min_eval` to the minimum of the current `min_eval` and the `eval` returned from the child's call.

3.  **Graph Construction**: The code then constructs the tree based on your description:
    *   Leaf nodes (D, E, F, G, H, I) are created first with arbitrary `value`s.
    *   Intermediate nodes (B, C) are created with their respective children.
    *   Finally, the root node (A) is created with B and C as its children.

4.  **Execution**: The `minimax` function is called with the root node `A` and `maximizing_player=True`, assuming the player at the root wants to maximize their outcome. The calculated `optimal_value` is then printed.
